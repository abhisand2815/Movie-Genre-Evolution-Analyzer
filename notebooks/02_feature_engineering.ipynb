Phase - 06 ----> Feature Engineering 

```python
import pandas as pd
import numpy as np

print("Feature engineering notebook is working!")
```

```python
data_path = "../data/processed/movies_integrated.csv"
df = pd.read_csv(data_path)
print("Dataset shape:", df.shape)
```

```python
df.head()
```

```python
print(df.columns.tolist())
```

```python
#creating Year 
df["year"] = df["startYear"].astype(int)
print(df["year"].min())
print(df["year"].max())
```

```python
#Creating Decade
df["decade"] = (df["year"] // 10) * 10
df[["primaryTitle", "year", "decade"]].head(10)
```

```python
def assign_era(year):
    if year < 1930:
        return "Silent Era"
    elif year < 1960:
        return "Golden Age"
    elif year < 1980:
        return "New Hollywood Era"
    elif year < 2000:
        return "Modern Era"
    else:
        return "Digital Era"


df["era"] = df["year"].apply(assign_era)

df[["primaryTitle", "year", "decade", "era"]].head(20)
```

```python
print("Movies by era:")
print(df["era"].value_counts())
```

```python
print("\nSample from each era:")

for era in df["era"].unique():
    print(f"\n--- {era} ---")
    print(
        df[df["era"] == era][
            ["primaryTitle", "year", "decade", "era"]
        ].head(3)
    )
```

```python
CURRENT_YEAR = 2026
df["movie_age"] = CURRENT_YEAR - df["year"]
df[["primaryTitle", "year", "movie_age"]].head(10)
```

```python
df["genre_count"] = df["genre_list"].apply(len)
df[
    ["primaryTitle", "genres", "genre_count"]
].head(20)
```

```python
# Fix genre_list
df["genre_list"] = df["genres"].str.split(",")
# Recalculate genre count
df["genre_count"] = df["genre_list"].apply(len)
df[["primaryTitle", "genres", "genre_list", "genre_count"]].head(20)
```

```python
print("Genre count distribution:")
print(df["genre_count"].value_counts().sort_index())
```

```python
print("Minimum genres per movie:", df["genre_count"].min())
print("Maximum genres per movie:", df["genre_count"].max())
```

```python
# Get all unique genres
all_genres = sorted(df["genre_list"].explode().unique())

print("Total unique genres:", len(all_genres))
print("\nGenres:")
print(all_genres)
```

```python
# Create binary indicator columns for each genre

for genre in all_genres:
    df[f"genre_{genre.lower().replace('-', '_').replace(' ', '_')}"] = (
        df["genre_list"].apply(lambda x: int(genre in x))
    )
print("Genre indicator columns created.")
```

```python
# Check actual genre indicator columns
indicator_columns = [
    f"genre_{genre.lower().replace('-', '_').replace(' ', '_')}"
    for genre in all_genres
]
print("Number of genre indicator columns:", len(indicator_columns))
print("\nGenre indicator columns:")
print(indicator_columns)
```

```python
df[
    [
        "primaryTitle",
        "genres",
        "genre_count",
        "genre_action",
        "genre_adventure",
        "genre_drama",
        "genre_comedy"
    ]
].head(20)
```

```python
# Create decade feature
df["decade"] = (df["year"] // 10) * 10
print("Decade feature created.")
print(df[["primaryTitle", "year", "decade"]].head(10))
```

```python
print(df[["primaryTitle", "year", "movie_age"]].head(10))
```

```python
# Create runtime categories
def classify_runtime(runtime):
    if pd.isna(runtime):
        return "Unknown"
    elif runtime < 90:
        return "Short"
    elif runtime <= 150:
        return "Medium"
    else:
        return "Long"

df["runtime_category"] = df["runtimeMinutes"].apply(classify_runtime)

print(df[["primaryTitle", "runtimeMinutes", "runtime_category"]].head(20))
```

```python
print("Runtime category distribution:")
print(df["runtime_category"].value_counts())
```

```python
print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
print("Total columns:", len(df.columns))

for i, col in enumerate(df.columns, start=1):
    print(i, col)
```

```python
print("===== FEATURE ENGINEERING CHECK =====")

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nMissing values:")
print(df.isna().sum())

print("\nRuntime categories:")
print(df["runtime_category"].value_counts())

print("\nGenre count:")
print(df["genre_count"].value_counts().sort_index())
```

```python
# Create rating categories
def classify_rating(rating):
    if pd.isna(rating):
        return "Unrated"
    elif rating < 5:
        return "Low"
    elif rating < 7:
        return "Average"
    elif rating < 8:
        return "Good"
    else:
        return "Excellent"

df["rating_category"] = df["averageRating"].apply(classify_rating)
print(df[["primaryTitle", "averageRating", "rating_category"]].head(20))
```

```python
print("Rating category distribution:")
print(df["rating_category"].value_counts())
```

```python
import numpy as np
# Log-transformed vote count
df["log_votes"] = np.log1p(df["numVotes"])

print(df[["primaryTitle", "numVotes", "log_votes"]].head(20))
```

```python
# Create popularity categories based on number of votes
def classify_popularity(votes):
    if pd.isna(votes):
        return "Unknown"
    elif votes < 100:
        return "Low"
    elif votes < 1000:
        return "Moderate"
    elif votes < 10000:
        return "Popular"
    else:
        return "Highly Popular"

df["popularity_category"] = df["numVotes"].apply(classify_popularity)
print(
    df[
        ["primaryTitle", "numVotes", "popularity_category"]
    ].head(20)
)
```

```python
print("Popularity distribution:")
print(df["popularity_category"].value_counts())
```

```python
# Rating weighted by log of votes
df["rating_popularity_score"] = (
    df["averageRating"] * df["log_votes"]
)
print(
    df[
        [
            "primaryTitle",
            "averageRating",
            "numVotes",
            "rating_popularity_score"
        ]
    ].head(20)
)
```

```python
new_features = [
    "decade",
    "era",
    "movie_age",
    "genre_count",
    "runtime_category",
    "rating_category",
    "log_votes",
    "popularity_category",
    "rating_popularity_score"
]

print("New engineered features:")
print(new_features)

print("\nFeature data:")
print(df[new_features].head(10))
```

```python
print("========== FINAL FEATURE ENGINEERING CHECK ==========")

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nNew feature missing values:")
print(df[new_features].isna().sum())

print("\nRating categories:")
print(df["rating_category"].value_counts())

print("\nPopularity categories:")
print(df["popularity_category"].value_counts())

print("\nRuntime categories:")
print(df["runtime_category"].value_counts())
```
